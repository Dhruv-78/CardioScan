```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

```

```python
#Loading 
import joblib
model = joblib.load('Utils/model.pkl')
label_encoder_sex = joblib.load('Utils/label_encoder_sex.pkl')
label_encoder_cpt = joblib.load('Utils/label_encoder_cpt.pkl')
label_encoder_restecg = joblib.load('Utils/label_encoder_restecg.pkl')
label_encoder_exang = joblib.load('Utils/label_encoder_exang.pkl')
label_encoder_st_slope = joblib.load('Utils/label_encoder_st_slope.pkl')
scaler = joblib.load('Utils/scaler.pkl')
boruta = joblib.load('Utils/boruta.pkl')    
pca = joblib.load('Utils/pca.pkl')
```

```python
#User data
age = 40
Sex = "M"
cpt = "ATA"
rBP = 140
chol = 289
fbs = 0
restecg = "Normal"
maxHR = 172
exang = "N"
oldpeak = 0
St_slope = "Up"

user_data = pd.DataFrame({
    'Age': [age],
    'Sex': [Sex],
    'ChestPainType': [cpt],
    'RestingBP': [rBP],
    'Cholesterol': [chol],
    'FastingBS': [fbs],
    'RestingECG': [restecg],
    'MaxHR': [maxHR],
    'ExerciseAngina': [exang],
    'Oldpeak': [oldpeak],
    'ST_Slope': [St_slope]
})
```

```python
#Preprocessing

#Encoding categorical variables
user_data['Sex'] = label_encoder_sex.transform(user_data['Sex'])
user_data['ChestPainType'] = label_encoder_cpt.transform(user_data['ChestPainType'])
user_data['ExerciseAngina'] = label_encoder_exang.transform(user_data['ExerciseAngina'])
user_data['ST_Slope'] = label_encoder_st_slope.transform(user_data['ST_Slope'])
user_data['RestingECG'] = label_encoder_restecg.transform(user_data['RestingECG'])

#  Scaling numerical features
user_data = scaler.transform(user_data)

#Feature selection
# user_data_selected = boruta.transform(user_data)

#Dimensionality reduction
user_data_pca = pca.transform(user_data)

#Prediction
prediction = model.predict(user_data_pca)
prediction_proba = model.predict_proba(user_data_pca)

print("Predicted class:", prediction[0])
print("Prediction probabilities:", prediction_proba[0])
```
